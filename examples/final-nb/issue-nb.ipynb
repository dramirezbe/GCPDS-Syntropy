```python
import numpy
import scipy
import matplotlib
import pandas
```

```python
sr = 2400000
dur = 0.01
t = numpy.linspace(0, dur, int(sr * dur))
iq = numpy.exp(1j * 2 * numpy.pi * 1000 * t)
iq = iq.astype(numpy.complex64)
```

```python
i = iq.real - numpy.mean(iq.real)
q = iq.imag - numpy.mean(iq.imag)
iq_bal = (i + 1j * q).astype(numpy.complex64)
```

```python
f, psd = scipy.signal.welch(iq_bal, fs=sr, nperseg=1024, noverlap=512)
```

```python
i2 = iq_bal.real - numpy.median(iq_bal.real)
q2 = iq_bal.imag - numpy.median(iq_bal.imag)
iq_clean = (i2 + 1j * q2).astype(numpy.complex64)
```

```python
f2, psd2 = scipy.signal.welch(iq_clean, fs=sr, nperseg=1024, noverlap=512)
```

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(f, 10 * numpy.log10(psd))
ax2.plot(f2, 10 * numpy.log10(psd2))
plt.show()
```
